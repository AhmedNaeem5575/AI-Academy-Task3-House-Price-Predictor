```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("housing.csv")
df.dropna(inplace=True)

# Create binary target for logistic regression
df['expensive'] = (df['median_house_value'] > 250000).astype(int)

# Normalize feature (median_income) and targets
X = df['median_income'].values
X = (X - X.mean()) / X.std()
X = X.reshape(-1, 1)

# For Linear Regression
y_lin = df['median_house_value'].values
y_lin = (y_lin - y_lin.mean()) / y_lin.std()
y_lin = y_lin.reshape(-1, 1)

# For Logistic Regression
y_log = df['expensive'].values.reshape(-1, 1)

# Hyperparameters
lr = 0.01
epochs = 1000
```

```python
w_lin = np.zeros((1, 1))
b_lin = 0

for _ in range(epochs):
    y_pred_lin = np.dot(X, w_lin) + b_lin
    error_lin = y_pred_lin - y_lin

    dw_lin = np.dot(X.T, error_lin) / len(X)
    db_lin = np.sum(error_lin) / len(X)

    w_lin -= lr * dw_lin
    b_lin -= lr * db_lin

print("Linear Regression:")
print("  Weight:", w_lin[0][0])
print("  Bias:", b_lin)
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

w_log = np.zeros((1, 1))
b_log = 0

for _ in range(epochs):
    z = np.dot(X, w_log) + b_log
    y_pred_log = sigmoid(z)
    error_log = y_pred_log - y_log

    dw_log = np.dot(X.T, error_log) / len(X)
    db_log = np.sum(error_log) / len(X)

    w_log -= lr * dw_log
    b_log -= lr * db_log

print("\nLogistic Regression:")
print("  Weight:", w_log[0][0])
print("  Bias:", b_log)
```

```python
import pickle

with open("linear_model.pkl", "wb") as f:
    pickle.dump((w_lin, b_lin), f)

with open("logistic_model.pkl", "wb") as f:
    pickle.dump((w_log, b_log), f)

print("\nModels saved as 'linear_model.pkl' and 'logistic_model.pkl'")
```
